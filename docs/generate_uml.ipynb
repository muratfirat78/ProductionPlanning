# Generate UML diagrams with Pyreverse

Uses `pyreverse` (bundled with `pylint`) and `graphviz` to statically analyze `src/` and render a class diagram and a package diagram to `docs/uml/`.

**Select the `uml-env` kernel** (created from `uml_env.yml`) before running this notebook.

cmd for creating the uml_env:
conda env create -f docs/uml_env.yml


```python
import subprocess
import shutil
from pathlib import Path

REPO_ROOT = Path.cwd().parent if Path.cwd().name == "docs" else Path.cwd()
SRC_DIR = REPO_ROOT / "src"
OUT_DIR = REPO_ROOT / "docs" / "uml"
OUT_DIR.mkdir(parents=True, exist_ok=True)

assert SRC_DIR.exists(), f"Expected source directory at {SRC_DIR}"
assert shutil.which("pyreverse") is not None, "pyreverse not found, activate the uml-env kernel/environment"
assert shutil.which("dot") is not None, "graphviz 'dot' binary not found, activate the uml-env kernel/environment"
```

```python
# -o png: render directly to PNG (requires graphviz)
# -d: output directory
# --source-roots: resolve the flat imports used by modules in src/

cmd = [
    "pyreverse",
    "-o", "png",
    "-d", str(OUT_DIR),
    "--source-roots", str(SRC_DIR),
    str(SRC_DIR),
]
result = subprocess.run(cmd, cwd=REPO_ROOT, capture_output=True, text=True)
print(result.stdout)
print(result.stderr)
result.check_returncode()
```

```python
# pyreverse names output files `classes.png` / `packages.png` by default
for name in ("classes.png", "packages.png"):
    path = OUT_DIR / name
    print(path, "exists:", path.exists())
```
